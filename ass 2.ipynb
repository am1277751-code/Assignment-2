```python


def filter_long_words(words_list, K):
    
    result = []
    
    
    for word in words_list:
        
        if len(word) > K:
        
            result.append(word)
            
    #
    return result

my_list = ["apple", "pie", "banana"]
K_value = 3


output = filter_long_words(my_list, K_value)

print(output)
```

```python

def area(shape, a, b=0):
    if shape == "circle":
        print(f"Circle Area: {3.14 * a * a}")
    elif shape == "rectangle":
        print(f"Rectangle Area: {a * b}")

area("circle", 5)
area("rectangle", 4, 6)
```

```python

class Restaurant:
    def __init__(self):
        self.menu_items = {}
        self.book_table = []
        self.customer_orders = []

    def add_item_to_menu(self, item, price):
        self.menu_items[item] = price

    def book_tables(self, table_number):
        self.book_table.append(table_number)

    def customer_order(self, order):
        self.customer_orders.append(order)

    def print_menu(self):
        return self.menu_items

    def print_reservations(self):
        return self.book_table

    def print_orders(self):
        return self.customer_orders


restaurant = Restaurant()

restaurant.add_item_to_menu("Burger", 10)
restaurant.add_item_to_menu("Pizza", 15)

restaurant.book_tables(1)
restaurant.book_tables(2)

restaurant.customer_order("Burger")
restaurant.customer_order("Pizza")

print(restaurant.print_menu())
print(restaurant.print_reservations())
print(restaurant.print_orders())
```
